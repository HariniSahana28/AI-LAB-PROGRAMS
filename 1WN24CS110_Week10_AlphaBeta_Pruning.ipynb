```python

def alpha_beta(node, alpha, beta, maximizing):
    global step
    step += 1

    # Leaf node
    if isinstance(node, int):
        print(f"Step {step}: Leaf value = {node}")
        return node

    if maximizing:
        value = float('-inf')
        print(f"\nMAX Node: alpha={alpha}, beta={beta}")

        for child in node:
            result = alpha_beta(child, alpha, beta, False)
            value = max(value, result)
            alpha = max(alpha, value)

            print(f"MAX: value={value}, alpha={alpha}, beta={beta}")

            if alpha >= beta:
                print("PRUNED: Remaining branches skipped!")
                break

        return value

    else:
        value = float('inf')
        print(f"\nMIN Node: alpha={alpha}, beta={beta}")

        for child in node:
            result = alpha_beta(child, alpha, beta, True)
            value = min(value, result)
            beta = min(beta, value)

            print(f"MIN: value={value}, alpha={alpha}, beta={beta}")

            if beta <= alpha:
                print("PRUNED: Remaining branches skipped!")
                break

        return value


# Game tree from PPT
tree = [
    [
        [10, 9],
        [14, 18]
    ],
    [
        [5, 4],
        [50, 3]
    ]
]

step = 0

print("ALPHA-BETA PRUNING\n")

result = alpha_beta(
    tree,
    float('-inf'),
    float('inf'),
    True
)

print("\nFinal Optimal Value =", result)
```
