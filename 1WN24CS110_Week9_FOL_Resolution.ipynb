```python

def resolution(c1, c2):
    for literal in c1:
        opposite = "~" + literal if not literal.startswith("~") else literal[1:]

        if opposite in c2:
            new_clause = (set(c1) | set(c2))
            new_clause.remove(literal)
            new_clause.remove(opposite)

            return list(new_clause), literal, opposite

    return None, None, None


def resolution_algorithm(clauses):
    iteration = 1

    print("INITIAL CLAUSES:")
    for i, c in enumerate(clauses):
        print(f"C{i+1} = {c}")

    while True:
        print(f"\n--- ITERATION {iteration} ---")
        new_clauses = []

        for i in range(len(clauses)):
            for j in range(i + 1, len(clauses)):

                result, l1, l2 = resolution(clauses[i], clauses[j])

                if result is not None:
                    print(f"\nResolving C{i+1} and C{j+1}")
                    print(f"Complementary literals: {l1} and {l2}")

                    if len(result) == 0:
                        print("Result = {} (EMPTY CLAUSE)")
                        print("\nCONCLUSION PROVED!")
                        return

                    print("New clause =", result)

                    if result not in clauses and result not in new_clauses:
                        new_clauses.append(result)

        if not new_clauses:
            print("\nNo more resolution possible.")
            print("Conclusion cannot be proved.")
            return

        clauses.extend(new_clauses)
        iteration += 1


# Input clauses
clauses = [
    ["P", "Q"],
    ["~P"],
    ["~Q"]
]

resolution_algorithm(clauses)
```
