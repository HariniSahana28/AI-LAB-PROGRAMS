```python


initial = (1, 2, 3,
           4, 0, 6,
           7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


def dls(state, path, depth):


    if state == goal:
        return path


    if depth == 0:
        return None

    zero = state.index(0)

    row = zero // 3
    col = zero % 3


    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:

        r = row + dr
        c = col + dc

        if 0 <= r < 3 and 0 <= c < 3:

            new_state = list(state)
            pos = r * 3 + c


            new_state[zero], new_state[pos] = \
                new_state[pos], new_state[zero]

            new_state = tuple(new_state)


            if new_state not in path:

                result = dls(
                    new_state,
                    path + [new_state],
                    depth - 1
                )

                if result:
                    return result

    return None



for depth in range(10):

    solution = dls(initial, [initial], depth)

    print("Depth:", depth)

    if solution:
        break



if solution:

    print("\nSolution using IDS:\n")

    for state in solution:

        print(state[0], state[1], state[2])
        print(state[3], state[4], state[5])
        print(state[6], state[7], state[8])
        print()

    print("Total moves:", len(solution) - 1)

else:
    print("No solution found")
```
