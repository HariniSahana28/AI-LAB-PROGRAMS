```python
import random
import math


def cost_function(board):
    n = len(board)
    cost = 0

    for i in range(n):
        for j in range(i + 1, n):

            # Same row
            if board[i] == board[j]:
                cost += 1

            # Same diagonal
            elif abs(board[i] - board[j]) == abs(i - j):
                cost += 1

    return cost


def generate_neighbor(board):
    n = len(board)
    neighbor = board.copy()


    col = random.randrange(n)


    new_row = random.randrange(n - 1)

    if new_row >= neighbor[col]:
        new_row += 1

    neighbor[col] = new_row

    return neighbor



def simulated_annealing(n=8):

    board = [random.randrange(n) for _ in range(n)]

    T = 100.0
    cooling_rate = 0.99
    min_temperature = 0.01
    max_steps = 1000


    current_cost = cost_function(board)

    print("Initial board:", board)
    print("Initial cost:", current_cost)


    while T > min_temperature and current_cost != 0:

        for _ in range(max_steps):

            if current_cost == 0:
                break


            neighbor = generate_neighbor(board)


            neighbor_cost = cost_function(neighbor)


            delta_E = neighbor_cost - current_cost

            if delta_E < 0:
                board = neighbor
                current_cost = neighbor_cost

            else:
                probability = math.exp(-delta_E / T)

                if random.random() < probability:
                    board = neighbor
                    current_cost = neighbor_cost


        T = T * cooling_rate


    if current_cost == 0:
        print("\nSolution found!")
        print("Final board:", board)
        print("Final cost:", current_cost)
    else:
        print("\nSolution not found.")
        print("Best board:", board)
        print("Final cost:", current_cost)

    return board


# Main program
simulated_annealing()
```
