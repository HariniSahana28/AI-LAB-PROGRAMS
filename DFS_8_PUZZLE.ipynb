```python
def dfs(start, goal):
    stack = [start]
    visited = set()

    while stack:
        current = stack.pop()

        if current == goal:
            print("Goal found!")
            print(current)
            return

        if tuple(current) in visited:
            continue

        visited.add(tuple(current))


        zero = current.index(0)
        row = zero // 3
        col = zero % 3


        moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

        for dr, dc in moves:
            r = row + dr
            c = col + dc

            if 0 <= r < 3 and 0 <= c < 3:
                new_state = current.copy()

                new_zero = r * 3 + c


                new_state[zero], new_state[new_zero] = \
                    new_state[new_zero], new_state[zero]

                if tuple(new_state) not in visited:
                    stack.append(new_state)

    print("No solution")


start = [1, 2, 3,
         4, 0, 6,
         7, 5, 8]

goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

dfs(start, goal)
```
